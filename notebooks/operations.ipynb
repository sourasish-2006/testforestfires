```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from pathlib import Path

filename = "Algerian_forest_fires_dataset_UPDATE.csv"
candidates = [
    Path(filename),
    Path("..") / filename,
    Path("..") / "dataset" / filename,
]

csv_path = next((path for path in candidates if path.is_file()), None)
if csv_path is None:
    raise FileNotFoundError(
        f"Could not find {filename}. Check the file location or update the path."
    )

dataset = pd.read_csv(csv_path, header=1)
dataset.head()
```

```python
dataset.info()
```

Data Cleaning


```python
#Missing values
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
df=dataset
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
df.head()
```

```python
df.isnull().sum()
```

```python
#Removing the null values
df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
#Remove the 122nd row
df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#fix spaces in column names
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

Changes the required columns as integer data type

```python
df = df.dropna().copy()
cols = ['month', 'day', 'year', 'Temperature', 'RH', 'Ws']
df[cols] = df[cols].apply(pd.to_numeric, errors='coerce')
df = df.dropna(subset=cols).copy()
df[cols] = df[cols].astype(int)
```

```python
df.info()
```

Changing the other columns to float data type


```python
objects=[features for features in df.columns if df[features].dtypes=='str']
```

```python
for i in objects:
    if i != 'Classes':
        df[i]=df[i].astype(float)
df.info()
```

```python
#Lets have the cleaned dataset
df.to_csv('Algerian_forest_fires_dataset_cleaned_dataset.csv',index=False)
```

Exploratory Data Analysis

```python
df_copy=df.head()
df_copy
```

```python
#Drop day, month and year
df_copy=df.drop(['day','month','year'],axis=1)
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
#Encoding of the categories in classes
labels = df_copy['Classes']
numeric = pd.to_numeric(labels, errors='coerce')
encoded = labels.astype('string').str.strip().str.lower().map({'not fire': 0, 'fire': 1})
df_copy['Classes'] = numeric.fillna(encoded)
df_copy.head()
```

```python
df_copy['Classes'].value_counts()
```

```python
#Plot density plot for all features
sns.set_theme()
df_copy.hist(bins=50, figsize=(20,15))
plt.show()
```

```python
#Percentage for pie chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
#plotting pie chart
classlabels = percentage.index.map({0: "Not Fire", 1: "Fire"})
plt.figure(figsize=(12,7))
plt.pie(percentage, labels=classlabels, autopct='%1.1f%%')
plt.title("Pie Chart of Classes")
plt.show()
```
